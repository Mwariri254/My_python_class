## step 1 import the random module
    - the 'random' module will shuffle the questions

```python
# Import the random module to shuffle the quiz questions
import random
```

### create the questions to be tested
    - each questions includes choices  and the correct answers

```python
# A list of dictionaries containing quiz questions, choices, and correct answers
questions = [
    {
        "question": "What is 5 + 3?",
        "choices": ["A. 6",
                   "B. 7",
                   "C. 8",
                   "D. 9"],
        "answer": "C"
    },
    {
        "question": "What language are we using?",
        "choices": ["A. Python",
                     "B. Java", 
                     "C. HTML", 
                     "D. CSS"],
        "answer": "A"
    },
    {
        "question": "What is the capital of France?",
        "choices": ["A. London",
                    "B. Paris",
                    "C. Rome",
                    "D. Madrid"],
        "answer": "B"
    },
    {
        "question": "Which data type stores True or False?",
        "choices": ["A. String", 
                    "B. Integer",
                    "C. Boolean",
                    "D. List"],
        "answer": "C"
    },
    {
        "question": "Which symbol is used for comments in Python?",
        "choices": ["A. //",
                    "B. #", 
                    "C. <!--", 
                    "D. **"],
        "answer": "B"
    }
]
```

## Step 3: Shuffle and Track Score

The questions are mixed into a random order, and the score starts at zero.

```python
# Randomly reorder the quiz questions
random.shuffle(questions)

# Initialize the score
score = 0
```

## Step 4: Display Questions

A `for` loop displays each question and its answer choices.

```python
# Loop through each quiz question``
for question in questions:
     # Display the question
    print("\n" + question["question"])
# Display all answer choices
    for choice in question["choices"]:
        print(choice)
  # Continue asking until the user enters a valid answer
    while True:
        try:
       # Get the user's answer and convert it to uppercase to ensure consistency
            answer = input("Enter A, B, C, or D: ").upper()
  # Check whether the answer is valid (A,B,C,D) and raise a valueerror if not
            if answer not in ["A", "B", "C", "D"]:
                raise ValueError

            break
        except ValueError:
    # Display an error message for invalid input
            print("Invalid input. Please enter A, B, C, or D.")
    # Compare the user's answer with the correct answer
    if answer == question["answer"]:
         # Display a success message and increase the score
        print("Correct!")
        score += 1
    # Display a message for an incorrect answer
    else:
        print("Incorrect.")
```

## Step 5: Display the Final Score

The final score is displayed, followed by feedback based on the result.

```python
# Display the user's final score
print(f"\nYour score is {score}/{len(questions)}.")

# Provide feedback based on the final score
if score == len(questions):
    print("Excellent!")
elif score >= len(questions) // 2:
    print("Good!")
else:
    print("Try Again!")
```
